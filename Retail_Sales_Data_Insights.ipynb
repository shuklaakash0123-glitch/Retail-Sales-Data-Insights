```python
import pandas as pd
from io import StringIO

csv_data = """Order_ID,Order_Date,Customer,Product,Category,Quantity,Unit_Price,Region,Payment_Mode
1001,2025-01-05,Rahul,Laptop,Electronics,1,55000,North,UPI
1002,2025-01-07,Priya,Mouse,Accessories,3,800,East,Cash
1003,2025-01-12,Amit,Keyboard,Accessories,2,1200,North,Card
1004,2025-02-03,Neha,Monitor,Electronics,1,15000,West,UPI
1005,2025-02-10,Rahul,Laptop,Electronics,1,55000,North,Card
1006,2025-02-15,Sanjay,Headphones,Accessories,2,2500,South,UPI
1007,2025-03-02,Priya,Mouse,Accessories,4,800,East,Cash
1008,2025-03-08,Amit,Monitor,Electronics,1,15000,West,Card
1009,2025-03-15,Neha,Keyboard,Accessories,2,1200,North,UPI
1010,2025-03-20,Rahul,Laptop,Electronics,1,55000,North,UPI
"""

df = pd.read_csv(StringIO(csv_data))
df
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df.dtypes
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df.dtypes
```

```python
df["Sales"] = df["Quantity"] * df["Unit_Price"]

df[["Order_ID", "Product", "Quantity", "Unit_Price", "Sales"]]
```

```python
df["Customer"] = df["Customer"].str.strip()
df["Product"] = df["Product"].str.strip()
df["Category"] = df["Category"].str.strip()
df["Region"] = df["Region"].str.strip()
df["Payment_Mode"] = df["Payment_Mode"].str.strip()

df.head()
```

```python
print("Customers:")
print(df["Customer"].unique())

print("\nCategories:")
print(df["Category"].unique())

print("\nRegions:")
print(df["Region"].unique())

print("\nPayment Modes:")
print(df["Payment_Mode"].unique())
```

```python
print("Quantity:")
print(df["Quantity"].describe())

print("\nUnit Price:")
print(df["Unit_Price"].describe())

print("\nSales:")
print(df["Sales"].describe())
```

```python
print("Maximum Quantity:", df["Quantity"].max())
print("Minimum Quantity:", df["Quantity"].min())

print("Maximum Unit Price:", df["Unit_Price"].max())
print("Minimum Unit Price:", df["Unit_Price"].min())

print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())
```

```python
total_sales = df["Sales"].sum()

print("Total Revenue:", total_sales)
```

```python
total_orders = df["Order_ID"].nunique()

print("Total Orders:", total_orders)
```

```python
average_order_value = df["Sales"].mean()

print("Average Order Value:", average_order_value)
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print(category_sales)
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

category_sales.plot(kind="bar")

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=0)
plt.show()
```

```python
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)

print(product_sales)
product_sales.head(5).plot(kind="bar")

plt.title("Top 5 Products by Revenue")
plt.xlabel("Product")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.show()
```

```python
df["Month"] = df["Order_Date"].dt.month_name()

monthly_sales = df.groupby("Month")["Sales"].sum()

print(monthly_sales)
```
